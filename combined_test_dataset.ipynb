```python
import pandas as pd
import numpy as np

# Path handles file locations
from pathlib import Path
```

```python
# where the raw dataset lives
dataset = Path(r"C:\shared\LSBU\ablation-study-with-developed-hypertuned-models-for-anomaly-detection-in-hrc-robotics\stochastic_PnP")

# the eight fault types, each in its own folder with a "test" subfolder inside
anomalies = ["push", "tcp_wrench", "drop_object", "force_stop",
             "outside_box", "change_weight", "long_wait", "scale_speed"]

# Collect every test file, remembering which fault it belongs to.
# Finding the files first, then reading them, keeps the two jobs separate
test_files = []

for anomaly in anomalies:
    folder = dataset / anomaly / "test"
    files = sorted(folder.glob("*.csv"))

    for file in files:
        test_files.append((anomaly, file))

    print(f"{anomaly}:{len(files)} runs")

print()
print("total runs:", len(test_files))
```

```python
# Every run gets read and added to this list, then joined at the end.
all_runs = []
run_number = 0

for anomaly in anomalies:

    for file in sorted((dataset / anomaly / "test").glob("*.csv")):

        # the recording. These files are separated by spaces, not commas.
        one_run = pd.read_csv(file, sep=r"\s+")

        # the answer key sits next to it, same name but _label.npy instead of .csv
        label = np.load(str(file).replace(".csv", "_label.npy"))

        one_run["label"] = (label > 0).astype(int)   # 0 and 32 become 0 and 1
        one_run["anomaly_type"] = anomaly
        one_run["run_number"] = run_number

        all_runs.append(one_run)
        run_number = run_number + 1

print("runs read:", len(all_runs))
```

```python
# Stack all 80 runs on top of each other.

test_data = pd.concat(all_runs, ignore_index=True)

print("rows   :", len(test_data))
print("columns:", test_data.shape[1])
```
