```python
import pandas as pd

# Path handles file locations. It reads more clearly than string paths
from pathlib import Path
```

```python
# where the raw dataset lives
dataset = Path(r"C:\shared\LSBU\ablation-study-with-developed-hypertuned-models-for-anomaly-detection-in-hrc-robotics\stochastic_PnP")

# the two folders of fault-free recordings
normal_folder = dataset / "normal"          # 508 runs, direct route
branch_folder = dataset / "normal_branch"   # 102 runs, they enter the program branch

# list the CSV files in each. sorted() keeps the order the same every time,
# so re-running this notebook always produces an identical file.
normal_files = sorted(normal_folder.glob("*.csv"))
branch_files = sorted(branch_folder.glob("*.csv"))

print("normal runs:", len(normal_files))
print("branch runs:", len(branch_files))
print("total      :", len(normal_files) + len(branch_files))
```

```python
# Every run gets read and added to this list, then joined at the end.
all_runs = []

# run_number labels which recording each row came from. The runs get stacked on
# top of each other, so without this there is no way to tell where one ends and
# the next begins.
run_number = 0

for file in normal_files + branch_files:

    one_run = pd.read_csv(file, sep=r"\s+")
    one_run["run_number"] = run_number

    all_runs.append(one_run)
    run_number = run_number + 1

print("runs read:", len(all_runs))
```

```python
# Stack all 610 runs on top of each other.

data = pd.concat(all_runs, ignore_index=True)

print("rows   :", len(data))
print("columns:", data.shape[1])
```

```python
# Show all columns.
pd.set_option("display.max_columns", None)

# Display the first 3,000 rows.
data.tail()
```

```python
# checking a few things before saving this file

print("runs found      :", data["run_number"].nunique())
print("missing values  :", data.isna().sum().sum())
print("non-numeric cols:", (data.dtypes == "object").sum())
print()
print("rows per run - shortest:", data["run_number"].value_counts().min())
print("rows per run - longest :", data["run_number"].value_counts().max())
```

```python
output = Path(r"C:\shared\LSBU\ablation-study-with-developed-hypertuned-models-for-anomaly-detection-in-hrc-robotics\combined_dataset.csv")

data.to_csv(output, index=False)

print("saved:", output.name, data.shape)
```
